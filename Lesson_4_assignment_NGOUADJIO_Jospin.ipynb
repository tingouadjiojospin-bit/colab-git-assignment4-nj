```python
# Python and Pandas Assessment Notebook

import pandas as pd
import numpy as np
from functools import reduce
```

```python
# Task 1: Create a DataFrame using Pandas
# Create a DataFrame with the following data:
# Name: Alice, Bob, Charlie, David
# Age: 25, 30, 35, 28
# City: New York, San Francisco, Los Angeles, Chicago

# Your code here:
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'Age': [25, 30, 35, 28],
    'City': ['New York', 'San Francisco', 'Los Angeles', 'Chicago']
}
df = pd.DataFrame(data)

print(df)
print("\nColumns:", list(df.columns))
print("\nData types:")
print(df.dtypes)   # Age is an integer column, Name and City are text
```

```python
# Task 2: Row and Column Manipulation
# Drop the 'City' column from the DataFrame created in Task 1

# Your code here:
# drop() returns a new DataFrame, so the original df is not changed
if 'City' in df.columns:
    df_no_city = df.drop(columns=['City'])
else:
    print("The 'City' column does not exist.")
    df_no_city = df.copy()

print(df_no_city)

# Check that only 'Name' and 'Age' are left
print("\nColumns left:", list(df_no_city.columns))
print("Only Name and Age?", list(df_no_city.columns) == ['Name', 'Age'])
```

```python
# Task 3: Handling Null Values
# Create a new DataFrame with some null values and fill them

# Your code here:
df_null = pd.DataFrame({
    'Name': ['Alice', None, 'Charlie', 'David'],
    'Age': [25, np.nan, 35, 28],
    'Score': [88.5, 92.0, np.nan, 75.0]
})
print("Original DataFrame:")
print(df_null)
print("\nNumber of null values per column:")
print(df_null.isnull().sum())

# Method 1: fill with fixed values (0 for numbers, 'Unknown' for text)
df_filled = df_null.fillna({'Name': 'Unknown', 'Age': 0, 'Score': 0})
print("\nMethod 1 - fillna with fixed values:")
print(df_filled)

# Method 2: fill numeric columns with the mean of the column
df_mean = df_null.copy()
df_mean['Age'] = df_mean['Age'].fillna(df_mean['Age'].mean())
df_mean['Score'] = df_mean['Score'].fillna(df_mean['Score'].mean())
df_mean['Name'] = df_mean['Name'].fillna('Unknown')
print("\nMethod 2 - fill with the column mean:")
print(df_mean)

# Method 3: drop the rows that contain a null value
df_dropped = df_null.dropna()
print("\nMethod 3 - dropna (remove rows with nulls):")
print(df_dropped)

print("\nNulls left after Method 1:", df_filled.isnull().sum().sum())
```

```python
# Task 4: GroupBy and Describe
# Using the following DataFrame, group by 'Category' and describe the 'Value' column

df_group = pd.DataFrame({
    'Category': ['A', 'B', 'A', 'B', 'A', 'C'],
    'Value': [10, 20, 15, 25, 30, 35]
})

# Your code here:
stats = df_group.groupby('Category')['Value'].describe()
print(stats)

# Interpretation of the results
print("\nInterpretation:")
for category, row in stats.iterrows():
    print(f"Category {category}: {int(row['count'])} value(s), mean = {row['mean']:.2f}, "
          f"min = {row['min']}, max = {row['max']}")
print("- A has the most values (3) and the widest spread (10 to 30).")
print("- B has 2 values with a mean of 22.5.")
print("- C has only 1 value (35), so its std is NaN: a standard deviation needs at least 2 values.")
```

```python
# Task 5: Concatenation and Merging
# Concatenate two DataFrames vertically and merge them horizontally

df1 = pd.DataFrame({'A': [1, 2], 'B': [3, 4]})
df2 = pd.DataFrame({'A': [5, 6], 'B': [7, 8]})
df3 = pd.DataFrame({'C': [9, 10], 'D': [11, 12]})

# Your code here:
# Vertical concatenation: df2 goes under df1 (ignore_index gives a new index 0..3)
df_vertical = pd.concat([df1, df2], ignore_index=True)
print("Vertical concatenation (df1 + df2):")
print(df_vertical)
print("Shape:", df_vertical.shape)

# Horizontal merge with df3: df3 has only 2 rows, so the last 2 rows get NaN
df_merged = pd.merge(df_vertical, df3, left_index=True, right_index=True, how='left')
print("\nMerged with df3 horizontally (on the index):")
print(df_merged)
print("Shape:", df_merged.shape)   # 4 rows and 4 columns: A, B, C, D
```

```python
# Task 6: Tuples and Sets
# Create a tuple of fruits and a set of numbers.
# Then, try to add an element to each and observe the difference.

# Your code here:
fruits = ('apple', 'banana', 'orange')
numbers_set = {1, 2, 3, 4, 5}
print("Tuple:", fruits)
print("Set:", numbers_set)

# A tuple is immutable: it has no add() or append() method
try:
    fruits.append('mango')
except AttributeError as e:
    print("\nError when adding to the tuple:", e)

# The only way is to create a new tuple
new_fruits = fruits + ('mango',)
print("New tuple created instead:", new_fruits)

# A set is mutable: add() works
numbers_set.add(6)
numbers_set.add(3)   # 3 is already there, sets keep only unique values
print("\nSet after adding 6 and 3:", numbers_set)

print("\nDifference: a tuple cannot be changed after it is created (immutable),")
print("while a set can be changed with add() and keeps only unique values.")
```

```python
# Task 8: Functions and Lambda
# Create a function to calculate the square of a number.
# Then, use a lambda function to do the same.

# Your code here:
def square(x):
    # Return the square of a number, refuse anything that is not a number
    if not isinstance(x, (int, float)):
        raise TypeError("square() needs a number")
    return x * x

square_lambda = lambda x: x * x

for value in [4, -3, 2.5]:
    print(f"square({value}) = {square(value)} | lambda({value}) = {square_lambda(value)}")

try:
    square("five")
except TypeError as e:
    print("Error:", e)
```

```python
# Task 9: Iterators and Generators
# Create an iterator for the first 5 even numbers.
# Then, create a generator for the same.

# Your code here:
class EvenIterator:
    # Iterator class that returns the first n even numbers (0, 2, 4, ...)
    def __init__(self, n=5):
        self.n = n
        self.count = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.count >= self.n:
            raise StopIteration       # tells the for loop to stop
        value = self.count * 2
        self.count += 1
        return value

def even_generator(n=5):
    # Generator function: yield gives one value at a time
    for i in range(n):
        yield i * 2

print("Iterator:", list(EvenIterator()))
print("Generator:", list(even_generator()))

# Using next() step by step
gen = even_generator()
print("next():", next(gen), next(gen), next(gen))
```

```python
# Task 10: Map, Reduce, and Filter
# Use map to square all numbers in a list.
# Use reduce to find the product of all numbers in a list.
# Use filter to get only even numbers from a list.

numbers = [1, 2, 3, 4, 5]

# Your code here:
squares = list(map(lambda x: x ** 2, numbers))
print("map - squares:", squares)

# reduce on an empty list would raise an error, so we give a start value of 1
product = reduce(lambda a, b: a * b, numbers, 1)
print("reduce - product:", product)

evens = list(filter(lambda x: x % 2 == 0, numbers))
print("filter - even numbers:", evens)
```

```python
# Task 11: Object-Oriented Programming - Creating a Class
# Create a class called 'Rectangle' with attributes 'length' and 'width'.
# Include methods to calculate area and perimeter.

# Your code here:
class Rectangle:
    def __init__(self, length, width):
        # A rectangle cannot have a negative or zero side
        if length <= 0 or width <= 0:
            raise ValueError("length and width must be positive")
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

    def perimeter(self):
        return 2 * (self.length + self.width)

r1 = Rectangle(5, 3)
r2 = Rectangle(10, 4.5)
for r in [r1, r2]:
    print(f"Rectangle {r.length} x {r.width}: area = {r.area()}, perimeter = {r.perimeter()}")

try:
    Rectangle(-2, 3)
except ValueError as e:
    print("Error:", e)
```
